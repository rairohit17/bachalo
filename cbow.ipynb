```python
import numpy as np 
from tensorflow import keras 
import matplotlib.pyplot as mlt 
from tensorflow.keras import layers 
```

```python
words=["I","love","machine","learning"]
```

```python
# X = np.array([
#     [1, 3],
#     [0, 2]
# ])
# y = np.array([2, 1])
# Step 2: Convert words into numbers
word_to_id = {}

for i in range(len(words)):
    word_to_id[words[i]] = i

print("Word Mapping:", word_to_id)

# Step 3: Create context and target
X = []
y = []

for i in range(1, len(words) - 1):
    context = [word_to_id[words[i - 1]],
               word_to_id[words[i + 1]]]

    target = word_to_id[words[i]]

    X.append(context)
    y.append(target)

X = np.array(X)
y = np.array(y)

print("Input:", X)
print("Target:", y)

```

```python
# Step 4: Create CBOW model
model = keras.Sequential([
    layers.Embedding(input_dim=len(words), output_dim=8),
    layers.GlobalAveragePooling1D(),
    layers.Dense(len(words), activation="softmax")
])

# Step 5: Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy"
)

# Step 6: Train model
model.fit(X, y, epochs=100)

# Step 7: Predict missing word
test = np.array([[
    word_to_id["I"],
    word_to_id["machine"]
]])

result = model.predict(test)

predicted_id = np.argmax(result)

print("Predicted Word:", words[predicted_id])
```
